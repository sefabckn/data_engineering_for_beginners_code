```python
import requests
import json
from pprint import pp as print ## prettier library for readble json output of response
```

```python
url = 'https://jsonplaceholder.typicode.com/posts'
response = requests.get(url=url)
print(response.status_code) # 200 means success
```

```python
print(response.content[:100])
```

```python
resp_json_data = response.json()
print(resp_json_data[:3])
```

```python
post_id = 1
response = requests.get(url=f'{url}/{post_id}') # to get a specific post by making the url dynamic
print(response.json())
```

```python
entity ='comments'
response = requests.get(url = f'{url}/{post_id}/{entity}')
print(response.json())
```

```python
# search for electronics on google
url = 'https://www.google.com/search'
query_params = {'q': 'electronics'}
response = requests.get(url=f'{url}', params=query_params)
response.content[:100] # This is a HTML file
```

```python
params = {'offset': 5, 'limit': 3}
url = 'https://pokeapi.co/api/v2/'
entity = 'ability'
response = requests.get(url = f'{url}{entity}', params=params)
response.json()
```
