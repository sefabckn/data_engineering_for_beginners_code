```python
orders = [
    {"customer_id": "C001", "product": "laptop", "quantity": 2, "price": 1200.00},
    {"customer_id": "C002", "product": "mouse", "quantity": 1, "price": 25.99},
    {"customer_id": "C001", "product": "keyboard", "quantity": 1, "price": 89.50},
    {"customer_id": "C003", "product": "monitor", "quantity": 1, "price": 299.99},
    {"customer_id": "C002", "product": "laptop", "quantity": 1, "price": 1200.00},
    {"customer_id": "C004", "product": "headphones", "quantity": 3, "price": 79.99},
    {"customer_id": "C001", "product": "webcam", "quantity": 1, "price": 45.00},
    {"customer_id": "C003", "product": "mouse", "quantity": 2, "price": 25.99},
    {"customer_id": "C002", "product": "speaker", "quantity": 1, "price": 150.00},
    {"customer_id": "C005", "product": "tablet", "quantity": 1, "price": 399.99}
]
```

'''
Customer Order Analysis: Write python code that processes a list of customer orders 
to calculate the total revenue and find top 3 the most frequent customer.

'''

```python
total_revenue = 0
top_customers = {}
counter = 0
for order in orders:
    total_revenue += order['quantity'] * order['price']
    if order['customer_id'] not in top_customers:
        top_customers[order['customer_id']] = 1
    else:
        top_customers[order['customer_id']] += 1


print(f'Total revenue: {total_revenue}', end='\n')
print('Top 3 most frequent customers are:', end='\n')

top_3 = sorted(top_customers.items(), key = lambda x: x[1], reverse = True,)[:3]

for customer, count in top_3:
    print(customer)
```

```python
email_list = [
    "john.doe@company.com",
    "jane.smith@email.co.uk",
    "invalid-email",
    "bob@gmail.com",
    "alice.brown@company.com",
    "john.doe@company.com",  # duplicate
    "missing@domain",
    "test@example.org",
    "@nodomain.com",
    "jane.smith@email.co.uk",  # duplicate
    "valid.user@site.net",
    "no-at-symbol.com",
    "another@test.io"
]
```

```python
def check_emails(email_list):
    valid_emails = []
    invalid_emails = []

    for email in email_list:

        # Rule 1: must contain @
        if '@' not in email:
            invalid_emails.append(email)
            continue

        # Rule 2: must contain .
        if '.' not in email:
            invalid_emails.append(email)
            continue

        at_index = email.index('@')
        dot_index = email.index('.')

        # Rule 2: . must be after @
        if dot_index <= at_index:
            invalid_emails.append(email)
            continue

        # Rule 3: must contain text before @
        if at_index == 0:
            invalid_emails.append(email)
            continue

        valid_emails.append(email)

    return {
        'valid_emails': valid_emails,
        'invalid_emails': invalid_emails
    }
```

```python
check_emails(email_list)
```

```python
sales_data = {
    "Alice Johnson": [15000, 18000, 22000, 16000, 19000, 21000],
    "Bob Smith": [12000, 14000, 11000, 13000, 15000, 16000],
    "Carol Davis": [25000, 28000, 30000, 27000, 32000, 29000]
}
```

```python
class SalesRep:
    def __init__(self, name: str, sales_amounts: list = None):
        self.name = name
        self.sales_amounts = sales_amounts if sales_amounts is not None else []

    def add_sale(self, amount: float):
        self.sales_amounts.append(amount)
        return self.sales_amounts

    def calculate_average(self):
        if not self.sales_amounts:
            return 0
        return sum(self.sales_amounts) / len(self.sales_amounts)

    def hit_target(self, target: float) -> bool:
        return self.calculate_average() >= target
reps = [SalesRep(name, amounts) for name, amounts in sales_data.items()]
for rep in reps:
    print(f"{rep.name}: avg = {rep.calculate_average():.2f}, hit target (20000)? {rep.hit_target(20000)}")
```
