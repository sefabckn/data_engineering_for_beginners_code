```python
%%sql
SELECT
  n.n_name AS nation_c_name,
  s.quantity AS supplied_items_quantity,
  c.quantity AS purchased_items_quantity
FROM
  nation n
  LEFT JOIN (
    SELECT
      n.n_nationkey,
      SUM(l.l_quantity) AS quantity
    FROM
      lineitem l
      JOIN supplier s ON l.l_suppkey = s.s_suppkey
      JOIN nation n ON s.s_nationkey = n.n_nationkey
    GROUP BY
      n.n_nationkey
  ) s ON n.n_nationkey = s.n_nationkey
  LEFT JOIN (
    SELECT
      n.n_nationkey,
      SUM(l.l_quantity) AS quantity
    FROM
      lineitem l
      JOIN orders o ON l.l_orderkey = o.o_orderkey
      JOIN customer c ON o.o_custkey = c.c_custkey
      JOIN nation n ON c.c_nationkey = n.n_nationkey
    GROUP BY
      n.n_nationkey
  ) c ON n.n_nationkey = c.n_nationkey;
```

```python
%%sql
SELECT
    o.o_orderkey,
    o.o_orderdate,
    COALESCE(l.l_orderkey, -1) AS lineitem_orderkey,
    l.l_shipdate
FROM
    orders o
    LEFT JOIN lineitem l ON o.o_orderkey = l.l_orderkey
    AND o.o_orderdate BETWEEN l.l_shipdate - INTERVAL '5' DAY
    AND l.l_shipdate + INTERVAL '5' DAY
LIMIT
    10;
```

```python
%%sql --Write a query that shows the number of items returned for each region name

SELECT
    r.r_name as region_name,
    count(*) as count_of_region
FROM region as r
LEFT JOIN nation as n on r.r_regionkey = n.n_regionkey
LEFT JOIN supplier as s on s.s_nationkey= n.n_nationkey
LEFT JOIN lineitem as l on l.l_suppkey = s.s_suppkey
GROUP BY r.r_name
```

```python
%%sql --List the top 10 most selling parts (part name)

SELECT
    p.p_name as part_name,
    SUM(l.l_quantity) as total_sold
FROM part as p
INNER JOIN lineitem as l on p.p_partkey = l.l_partkey
GROUP BY p_partkey, p_name
ORDER BY total_sold DESC
LIMIT 10;
```

```python
%%sql -- Sellers (name) who have sold at least one of the top 10 selling parts
SELECT DISTINCT sup.s_name as supplier
FROM supplier as  sup
INNER JOIN partsupp as pupp on pupp.ps_suppkey = sup.s_suppkey
INNER JOIN
(
    SELECT 
        p.p_partkey,
        p.p_name as part_name,
        sum(l.l_quantity) as top_sells
    FROM lineitem as l
    INNER JOIN part as p on p.p_partkey = l.l_partkey
    GROUP BY p.p_partkey, p.p_name
    ORDER BY top_sells DESC
) as top_parts on top_parts.p_partkey = pupp.ps_partkey
LIMIT 10;
```

```python
%%sql

SELECT DISTINCT
    s.s_name AS seller_name
FROM lineitem AS l
INNER JOIN supplier AS s
    ON l.l_suppkey = s.s_suppkey
INNER JOIN (
    SELECT
        p.p_partkey
    FROM lineitem AS l
    INNER JOIN part AS p
        ON l.l_partkey = p.p_partkey
    GROUP BY
        p.p_partkey
) AS t
    ON l.l_partkey = t.p_partkey
LIMIT 10;
```

```python
%%sql -- Number of items returned for each order price bucket. The definition of order price bucket is shown below.

SELECT count(*) as num_of_returns,
    CASE
        WHEN o.o_totalprice > 100000 THEN 'high'
        WHEN o.o_totalprice BETWEEN 25000 AND 100000 THEN 'medium'
    ELSE 'low'
END AS order_price_bucket
FROM lineitem as l
INNER JOIN orders as o on o.o_orderkey = l.l_orderkey
WHERE l.l_returnflag = 'R'
GROUP BY CASE
        WHEN o.o_totalprice > 100000 THEN 'high'
        WHEN o.o_totalprice BETWEEN 25000 AND 100000 THEN 'medium'
    ELSE 'low'
END
ORDER BY order_price_bucket ASC;



    
```

```python
%%sql -- Average time (in days) between receiptdate and shipdate for each nation (name)
SELECT n.n_name as nation,
       ROUND(AVG(date_diff(l.l_receiptdate, l.l_shipdate)), 2) as avg_delivery_days
FROM lineitem as l 
INNER JOIN orders o on o.o_orderkey = l.l_orderkey
INNER JOIN customer c on c.c_custkey = o.o_custkey
INNER JOIN nation n on n.n_nationkey = c.c_nationkey
WHERE l.l_shipdate is not null  and l.l_receiptdate is not null 
GROUP BY n.n_name
ORDER BY avg_delivery_days
LIMIT 10;
```

```python
%%sql

SELECT
    n.n_name AS nation_name,
    ROUND(AVG(DATEDIFF(l.l_receiptdate, l.l_shipdate)), 2) AS avg_delivery_days
FROM lineitem AS l
INNER JOIN orders AS o
    ON l.l_orderkey = o.o_orderkey
INNER JOIN customer AS c
    ON o.o_custkey = c.c_custkey
INNER JOIN nation AS n
    ON c.c_nationkey = n.n_nationkey
WHERE l.l_receiptdate IS NOT NULL
  AND l.l_shipdate IS NOT NULL
GROUP BY
    n.n_name
ORDER BY
    avg_delivery_days ASC
LIMIT 10;
```

```python
%%sql -- find the amount of money lost on discounts, Discounted price is in lineitem table called extendedprice, totalprice from orders table includes the discounted price, so difference between those columns give us total lost amount
with lineitem_agg as (
    SELECT
        l_orderkey,
        l_extendedprice,
        l_discount,
        
        SUM(l_extendedprice) as total_price_without_discount
    FROM lineitem
    GROUP BY l_orderkey, l_extendedprice
)
SELECT
    o.o_orderkey,
    l.l_extendedprice,
    o.o_totalprice,
    l.total_price_without_discount - o.o_totalprice as amount_lost_to_discount
FROM orders as o
JOIN lineitem_agg l on l.l_orderkey = o.o_orderkey
ORDER BY o.o_orderkey;
```

```python
%%sql
with discounted_price as 
(
    SELECT l_orderkey,
        (1 - l_discount) * l_extendedprice as discounted_price
    FROM lineitem
)
SELECT o.o_totalprice,
       d.discounted_price
FROM orders as o
JOIN discounted_price as d on d.l_orderkey = o.o_orderkey
LIMIT 20;
```

```python
%%sql -- Sellers (name) who have sold at least one of the top 10 selling parts (use CTE)

with top_10 as (
    SELECT p.p_name,
           p.p_partkey,
           SUM(l.l_quantity) total_quantity
    FROM lineitem as l
    INNER JOIN part as p on p.p_partkey = l.l_partkey
    GROUP BY  p.p_partkey, p.p_name
    ORDER BY total_quantity DESC
    LIMIT 10
)
SELECT DISTINCT s.s_name as seller_name
FROM lineitem as l
INNER JOIN supplier as s on s.s_suppkey = l.l_suppkey
INNER JOIN top_10 as t on l.l_partkey = t.p_partkey
LIMIT 10;
```

```python
%%sql --  Write a query to calculate the daily running average of the total price for every customer.

SELECT
    o_custkey,
    o_orderdate,
    o_totalprice,
    ROUND(AVG(o_totalprice) OVER(PARTITION BY o_custkey ORDER BY o_orderdate), 2) as running_sum
FROM orders
WHERE o_custkey = 4
ORDER BY o_orderdate
LIMIT 10;
```

```python
%%sql
-- Let's look at an example showing the difference between RANK, DENSE_RANK and ROW_NUMBER
SELECT 
    order_date,
    order_id,
    total_price,
    ROW_NUMBER() OVER (PARTITION BY order_date ORDER BY total_price) AS row_number,
    RANK() OVER (PARTITION BY order_date ORDER BY total_price) AS rank,
    DENSE_RANK() OVER (PARTITION BY order_date ORDER BY total_price) AS dense_rank
FROM (
    SELECT 
        '2024-07-08' AS order_date, 'order_1' AS order_id, 100 AS total_price UNION ALL
    SELECT 
        '2024-07-08', 'order_2', 200 UNION ALL
    SELECT 
        '2024-07-08', 'order_3', 150 UNION ALL
    SELECT 
        '2024-07-08', 'order_4', 90 UNION ALL
    SELECT 
        '2024-07-08', 'order_5', 100 UNION ALL
    SELECT 
        '2024-07-08', 'order_6', 90 UNION ALL
    SELECT 
        '2024-07-08', 'order_7', 100 UNION ALL
    SELECT 
        '2024-07-10', 'order_8', 100 UNION ALL
    SELECT 
        '2024-07-10', 'order_9', 100 UNION ALL
    SELECT 
        '2024-07-10', 'order_10', 100 UNION ALL
    SELECT 
        '2024-07-11', 'order_11', 100
) AS orders
ORDER BY order_date, row_number;
```

```python
%%sql
SELECT
    order_month,
    o_custkey,
    total_price,
    ROUND(AVG(total_price) OVER(
        PARTITION BY o_custkey 
        ORDER BY order_month ROWS BETWEEN 1 PRECEDING 
        AND 1 FOLLOWING
        ), 2
    ) as three_mo_total_price_avg
FROM (
    SELECT
        date_format(o_orderdate, 'yyyy-MM') as order_month,
        o_custkey,
        SUM(o_totalprice) as total_price
    FROM orders
    GROUP BY 1, 2
    )
LIMIT 5;
```

```python
%%sql
SELECT
    order_month,
    o_custkey,

    ROUND(
        AVG(total_price) OVER (
            PARTITION BY o_custkey
            ORDER BY order_month
            ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
        ),
        2
    ) AS three_mo_total_price_avg,

    ROUND(
        AVG(total_price) OVER (
            PARTITION BY o_custkey
            ORDER BY order_month
            RANGE BETWEEN 2 PRECEDING AND CURRENT ROW
        ),
        2
    ) AS consecutive_three_mo_total_price_avg

FROM (
    SELECT
        CAST(DATE_TRUNC('MONTH', o_orderdate) AS DATE) AS order_month,
        o_custkey,
        SUM(o_totalprice) AS total_price
    FROM orders
    GROUP BY
        CAST(DATE_TRUNC('MONTH', o_orderdate) AS DATE),
        o_custkey
) t;
```

```python

%%sql

WITH monthly_orders AS (
    SELECT
        CAST(date_format(o_orderdate, 'yyyy-MM-01') AS DATE) AS order_month,
        o_custkey,
        SUM(o_totalprice) AS total_price
    FROM orders
    GROUP BY
        CAST(date_format(o_orderdate, 'yyyy-MM-01') AS DATE),
        o_custkey
)

SELECT
    order_month,
    o_custkey,
    total_price,

    AVG(total_price) OVER (
        PARTITION BY o_custkey
        ORDER BY order_month
        ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
    ) AS three_mo_total_price_avg,

    AVG(total_price) OVER (
        PARTITION BY o_custkey
        ORDER BY order_month
        RANGE BETWEEN INTERVAL 2 MONTHS PRECEDING AND CURRENT ROW
    ) AS consecutive_three_mo_total_price_avg

FROM monthly_orders

ORDER BY
    o_custkey,
    order_month

LIMIT 10;

```

```python
%%sql
WITH daily_spending AS (
    SELECT
        CAST(o_orderdate AS DATE) AS order_date,
        o_custkey,
        SUM(o_totalprice) AS daily_spending
    FROM orders
    GROUP BY
        CAST(o_orderdate AS DATE),
        o_custkey
),

ranked_customers AS (
    SELECT
        order_date,
        o_custkey,
        daily_spending,
        RANK() OVER (
            PARTITION BY order_date
            ORDER BY daily_spending ASC
        ) AS spending_rank
    FROM daily_spending
)

SELECT
    order_date,
    o_custkey,
    daily_spending,
    spending_rank
FROM ranked_customers
WHERE spending_rank <= 3
ORDER BY
    order_date,
    spending_rank
LIMIT 20;
```
